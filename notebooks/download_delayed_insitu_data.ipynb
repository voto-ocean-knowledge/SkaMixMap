```python
from pathlib import Path
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import pandas as pd
import requests
```

### Download glider data

```python
missions = ["SEA044_M109", "SEA068_M46", "SEA069_M52"]
dtype = "nrt"
dtype = "delayed"
glider_data = {}
data_dir = Path("data")
if not data_dir.exists():
    data_dir.mkdir()
for mission in missions:
    dataset_id = f"{dtype}_{mission}"
    outfile = data_dir / f"{dataset_id}.nc"
    if not outfile.exists():
        url_mission = f"https://erddap.observations.voiceoftheocean.org/erddap/files/{dataset_id}/mission_timeseries.nc"  
        response_glider = requests.get(url_mission)
        with open(outfile, "wb") as f:
            f.write(response_glider.content)
    glider_data[mission] = xr.open_dataset(outfile)
```
